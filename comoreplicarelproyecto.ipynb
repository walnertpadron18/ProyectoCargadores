Cómo replicar el proyecto: ¿Dónde invertir en carga pública en Europa?

1. Requisitos
Para replicar el proyecto necesitas:
• Python 3 (recomendado con Anaconda) y Jupyter o VS Code para abrir el notebook.
• MySQL Server y MySQL Workbench.
• Tableau (Desktop o Public) para abrir el dashboard.
• Conexión a internet, porque la población se descarga de la API de Eurostat.
Librerías de Python:
pip install pandas requests sqlalchemy pymysql pycountry

2. Estructura de carpetas
Todos los archivos deben estar en la misma carpeta del proyecto:

3. Pasos para replicarlo
Paso 1: crear la base de datos
En MySQL Workbench, ejecuta:
CREATE DATABASE coches_electricos;
Después, ejecuta el script SQL de creación de tablas, con las claves primarias y foráneas. Hazlo antes del notebook para que las tablas tengan los tipos y las claves correctos.
Paso 2: ejecutar el notebook
Abre Proyecto_final_cargadores.ipynb y ejecuta las celdas en orden. Al principio te pedirá la contraseña de MySQL. El notebook hace lo siguiente:
1.  Limpia las matriculaciones de Ember y deja solo los países de la UE.
2.  Crea las tablas paises, tipos_motor y matriculaciones, y comprueba que BEV + PHEV coincide con el total de eléctricos.
3.  Carga los puntos de carga (lenta, media y rápida).
4.  Descarga la población de Eurostat y se queda con el último año de cada país.
5.  Carga las ayudas públicas.
6.  Calcula la flota de eléctricos en circulación.
7.  Guarda las 7 tablas en MySQL.
Paso 3: comprobar los datos en MySQL
En Workbench, comprueba que las tablas tienen filas:
USE coches_electricos;
SELECT COUNT(*) FROM paises;
SELECT COUNT(*) FROM matriculaciones;
Paso 4: calcular las métricas
Ejecuta las consultas SQL de métricas (eléctricos por 1.000 habitantes, eléctricos por punto de carga, porcentaje de carga rápida, ratio eléctricos/gasolina, índice de apoyo e índice de oportunidad) y exporta los resultados a CSV.
Paso 5: abrir el dashboard
Abre el libro de Tableau. Si los CSV están en otra ruta, ve a Fuente de datos → clic derecho en la conexión → Editar conexión y selecciona los archivos nuevos.

4. Problemas frecuentes
Error
Causa
Solución
No module named 'cleaning'
cleaning.py no está en la carpeta del notebook
Mueve el archivo a esa carpeta y reinicia el kernel
Unknown database 'coches_electricos'
La base de datos no existe
Ejecuta el Paso 1
Datos duplicados en las tablas
El notebook se ejecutó dos veces (guarda con append)
Vacía las tablas con TRUNCATE o vuelve a crear la base antes de repetir
Faltan países en la población
Eurostat no tiene el año más reciente
El notebook usa automáticamente el último año disponible
